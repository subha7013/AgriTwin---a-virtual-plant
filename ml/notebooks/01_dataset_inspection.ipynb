# AgriTwin ML Pipeline: 01 - Dataset Acquisition and Inspection

## 1. Dataset Overview
Dataset: Tomato Leaf Disease Dataset (10 Classes)
Source: Kaggle (kaustubhb999/tomatoleaf)
Objective: Inspect raw files, verify directory integrity, check image formats, color modes, and detect corruptions.

```python
import os
from PIL import Image
from pathlib import Path

DATASET_PATH = Path("../dataset/raw")

class_counts = {}
resolutions = set()
formats = set()
modes = set()
corrupted_files = []

for class_dir in sorted(DATASET_PATH.iterdir()):
    if not class_dir.is_dir():
        continue
    img_files = [f for f in class_dir.iterdir() if f.suffix.lower() in [".jpg", ".jpeg", ".png"]]
    class_counts[class_dir.name] = len(img_files)
    
    for img_path in img_files:
        try:
            with Image.open(img_path) as img:
                resolutions.add(img.size)
                formats.add(img.format)
                modes.add(img.mode)
        except Exception as e:
            corrupted_files.append((str(img_path), str(e)))

print(f"Total Disease Classes: {len(class_counts)}")
print(f"Total Images: {sum(class_counts.values())}")
print(f"Corrupted Files Detected: {len(corrupted_files)}")
print(f"Image Resolutions: {resolutions}")
print(f"Image Formats: {formats}")
print(f"Color Modes: {modes}")
```
