# AgriTwin ML Pipeline: 03 - Data Preprocessing & Augmentation Pipeline

## 1. Stratified Splitting
70% Training | 15% Validation | 15% Independent Test Set
Avoid data leakage by isolating test samples prior to training.

```python
import os
import tensorflow as tf
from tensorflow.keras import layers

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    "../dataset/processed/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    "../dataset/processed/validation",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    "../dataset/processed/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

print("Class Names:", train_ds.class_names)
```

## 2. Training Data Augmentation
Geometric and illumination transforms applied ONLY to training samples to boost generalization.

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.12),
    layers.RandomZoom(0.08),
    layers.RandomContrast(0.1)
], name="data_augmentation")

import matplotlib.pyplot as plt
for images, _ in train_ds.take(1):
    first_img = images[0]
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(first_img.numpy().astype("uint8"))
    plt.title("Original Image")
    plt.axis("off")
    
    plt.subplot(1, 2, 2)
    augmented = data_augmentation(tf.expand_dims(first_img, 0))
    plt.imshow(augmented[0].numpy().astype("uint8"))
    plt.title("Augmented Image")
    plt.axis("off")
    plt.show()
```
